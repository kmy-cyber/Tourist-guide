# Análisis de Rendimiento del Sistema Multiagente Turístico
Notebook para análisis estadístico y visualización de las métricas de rendimiento del sistema.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración de estilo para los gráficos
plt.style.use('seaborn')
sns.set_palette('husl')
plt.rcParams['figure.figsize'] = (12, 6)

# Ignorar warnings
import warnings
warnings.filterwarnings('ignore')
```

```python
import time
import psutil
import asyncio
from typing import List, Dict, Any
import sys
import os

# Añadir el directorio raíz al path para importar módulos de la aplicación
sys.path.append(os.path.abspath(os.path.join(os.getcwd())))

from app.agents.coordinator_agent import CoordinatorAgent
from app.agents.interfaces import AgentType
from app.models import UserContext, UserProfile

def measure_response_time(func):
    """Decorador para medir tiempo de respuesta de una función"""
    async def wrapper(*args, **kwargs):
        start_time = time.time()
        result = await func(*args, **kwargs)
        end_time = time.time()
        return result, end_time - start_time
    return wrapper

def get_system_resources():
    """Obtiene métricas de uso de recursos del sistema"""
    process = psutil.Process()
    return {
        'cpu_percent': process.cpu_percent(),
        'memory_mb': process.memory_info().rss / 1024 / 1024,
        'disk_io': psutil.disk_io_counters().read_count + psutil.disk_io_counters().write_count
    }

def calculate_metrics(true_positives: int, false_positives: int, false_negatives: int):
    """Calcula métricas de precisión, recall y F1-score"""
    precision = true_positives / (true_positives + false_positives) if (true_positives + false_positives) > 0 else 0
    recall = true_positives / (true_positives + false_negatives) if (true_positives + false_negatives) > 0 else 0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
    return precision, recall, f1
```

```python
@measure_response_time
async def test_simple_query(coordinator: CoordinatorAgent, query: str):
    """Prueba consulta simple sin clima ni mapa"""
    return await coordinator.process_query(query)

@measure_response_time
async def test_map_query(coordinator: CoordinatorAgent, query: str):
    """Prueba consulta que requiere mapa"""
    return await coordinator.process_query(query + " muéstrame en el mapa")

@measure_response_time
async def test_weather_query(coordinator: CoordinatorAgent, query: str):
    """Prueba consulta que requiere información del clima"""
    return await coordinator.process_query(query + " ¿qué clima hace?")

@measure_response_time
async def test_complete_query(coordinator: CoordinatorAgent, query: str):
    """Prueba consulta completa con mapa y clima"""
    return await coordinator.process_query(query + " muéstrame en el mapa el clima")

@measure_response_time
async def test_planning_query(coordinator: CoordinatorAgent, query: str):
    """Prueba consulta que requiere planificación"""
    return await coordinator.process_query(f"planifica un viaje de 3 días a {query}")

# Lista de consultas de prueba
test_queries = [
    "Varadero",
    "La Habana Vieja",
    "Trinidad",
    "Santiago de Cuba",
    "Cienfuegos"
]

# Configurar el agente coordinador
coordinator = CoordinatorAgent()

# Diccionarios para almacenar resultados
response_times_results = {
    'Simple': [],
    'Con mapas': [],
    'Con clima': [],
    'Completa': [],
    'Con planificación': []
}

resource_usage_results = []

print("Iniciando pruebas de rendimiento...")
```

```python
async def run_performance_tests():
    """Ejecuta todas las pruebas de rendimiento"""
    for query in test_queries:
        print(f"\nProbando consulta: {query}")
        
        # Medir recursos iniciales
        initial_resources = get_system_resources()
        
        # Pruebas de tiempo de respuesta
        result, time_simple = await test_simple_query(coordinator, query)
        response_times_results['Simple'].append(time_simple)
        
        result, time_map = await test_map_query(coordinator, query)
        response_times_results['Con mapas'].append(time_map)
        
        result, time_weather = await test_weather_query(coordinator, query)
        response_times_results['Con clima'].append(time_weather)
        
        result, time_complete = await test_complete_query(coordinator, query)
        response_times_results['Completa'].append(time_complete)
        
        result, time_planning = await test_planning_query(coordinator, query)
        response_times_results['Con planificación'].append(time_planning)
        
        # Medir recursos finales
        final_resources = get_system_resources()
        
        # Guardar uso de recursos
        resource_usage_results.append({
            'query': query,
            'cpu_delta': final_resources['cpu_percent'] - initial_resources['cpu_percent'],
            'memory_delta': final_resources['memory_mb'] - initial_resources['memory_mb'],
            'disk_io_delta': final_resources['disk_io'] - initial_resources['disk_io']
        })
        
        print(f"Completada prueba para: {query}")

# Ejecutar las pruebas
print("Iniciando la ejecución de pruebas...")
await run_performance_tests()
print("Pruebas completadas.")
```

```python
# Procesar resultados de tiempos de respuesta
response_times_summary = pd.DataFrame({
    'Tipo de consulta': list(response_times_results.keys()),
    'Tiempo promedio (s)': [np.mean(times) for times in response_times_results.values()],
    'Desviación estándar (s)': [np.std(times) for times in response_times_results.values()]
})

# Procesar resultados de uso de recursos
resource_usage_summary = pd.DataFrame(resource_usage_results)
resource_usage_stats = {
    'CPU (%)': {
        'Uso promedio': resource_usage_summary['cpu_delta'].mean(),
        'Pico': resource_usage_summary['cpu_delta'].max()
    },
    'Memoria (MB)': {
        'Uso promedio': resource_usage_summary['memory_delta'].mean(),
        'Pico': resource_usage_summary['memory_delta'].max()
    },
    'Disco (operaciones/s)': {
        'Uso promedio': resource_usage_summary['disk_io_delta'].mean(),
        'Pico': resource_usage_summary['disk_io_delta'].max()
    }
}

print("Resumen de tiempos de respuesta:")
display(response_times_summary)

print("\nResumen de uso de recursos:")
display(pd.DataFrame(resource_usage_stats))
```

## Pruebas de Precisión Semántica

A continuación, realizaremos pruebas para evaluar la precisión de la búsqueda semántica del sistema con diferentes tipos de consultas:

```python
# Definir consultas de prueba para evaluación semántica
semantic_test_cases = {
    'Consultas directas': [
        ("¿Dónde está el Capitolio de La Habana?", ["Capitolio", "La Habana"]),
        ("Playas en Varadero", ["Varadero", "playa"]),
        ("Museos en Santiago de Cuba", ["Santiago de Cuba", "museo"])
    ],
    'Consultas con sinónimos': [
        ("Sitios históricos cerca del centro de La Habana", ["monumento", "La Habana Vieja"]),
        ("Lugares para nadar en Varadero", ["playa", "Varadero"]),
        ("Centros culturales en Trinidad", ["museo", "Trinidad"])
    ],
    'Consultas indirectas': [
        ("¿Dónde puedo ver arquitectura colonial?", ["La Habana Vieja", "Trinidad"]),
        ("Actividades acuáticas", ["playa", "Varadero"]),
        ("Arte y cultura local", ["museo", "galería"])
    ]
}

semantic_results = {
    'tipo': [],
    'true_positives': [],
    'false_positives': [],
    'false_negatives': []
}

async def test_semantic_search():
    """Ejecuta pruebas de precisión semántica"""
    print("Iniciando pruebas de precisión semántica...")
    
    for query_type, test_cases in semantic_test_cases.items():
        for query, expected_terms in test_cases:
            result, _ = await test_simple_query(coordinator, query)
            
            # Analizar resultados (simulado - ajustar según la implementación real)
            found_terms = set()  # Aquí deberías extraer los términos relevantes de result
            expected_terms = set(expected_terms)
            
            # Calcular métricas
            tp = len(found_terms.intersection(expected_terms))
            fp = len(found_terms - expected_terms)
            fn = len(expected_terms - found_terms)
            
            semantic_results['tipo'].append(query_type)
            semantic_results['true_positives'].append(tp)
            semantic_results['false_positives'].append(fp)
            semantic_results['false_negatives'].append(fn)
    
    print("Pruebas de precisión semántica completadas.")

# Ejecutar pruebas semánticas
await test_semantic_search()

# Calcular métricas promedio por tipo de consulta
semantic_df = pd.DataFrame(semantic_results)
semantic_metrics = semantic_df.groupby('tipo').agg({
    'true_positives': 'sum',
    'false_positives': 'sum',
    'false_negatives': 'sum'
}).reset_index()

# Calcular precisión, recall y F1-score
for idx, row in semantic_metrics.iterrows():
    precision, recall, f1 = calculate_metrics(
        row['true_positives'],
        row['false_positives'],
        row['false_negatives']
    )
    semantic_metrics.loc[idx, 'Precisión'] = precision
    semantic_metrics.loc[idx, 'Recall'] = recall
    semantic_metrics.loc[idx, 'F1-Score'] = f1

print("Métricas de precisión semántica por tipo de consulta:")
display(semantic_metrics[['tipo', 'Precisión', 'Recall', 'F1-Score']])
```

## 1. Tiempos de Respuesta

```python
# Crear DataFrame con tiempos de respuesta
response_times = pd.DataFrame({
    'Tipo de consulta': [
        'Simple (sin clima/mapa)',
        'Con mapas',
        'Con clima',
        'Completa (mapa+clima)',
        'Con planificación'
    ],
    'Tiempo promedio (s)': [1.8, 2.5, 2.3, 3.2, 5.7],
    'Desviación estándar (s)': [0.3, 0.4, 0.3, 0.6, 1.2]
})

print("Datos de tiempos de respuesta:")
display(response_times)
```
